```python
import pandas as pd
```

```python
salesman = pd.read_excel(r'C:\Users\keong\Desktop\IMTools\IMtool\IMtool app\master_files\master price.xlsx')
```

```python
def find_last_digit_index(s):
    for i, char in enumerate(reversed(s)):
        if char.isdigit():
            return len(s) - i
    return -1  # if no digit found

# Apply the function to each cell in the column
salesman['max_num'] = salesman['pur_code'].apply(find_last_digit_index)

# Extract characters starting from 'max_num' index
salesman["pur_code"] = salesman.apply(
    lambda row: row["pur_code"][row["max_num"] :], 
    axis=1
)
# salesman["pur_code"] = salesman["pur_code"].str.slice(salesman["max_num"])
salesman = salesman['pur_code']

salesman.drop_duplicates(inplace=True)
```

```python
import requests

# Define the token endpoint and your client credentials
token_url = 'https://preprod-api.myinvois.hasil.gov.my/connect/token/'
client_id = '5d751953-026b-4dc2-a754-2bd32b10b4bc'
client_secret = '892e4c8a-cd23-4ccc-b821-f03d1e1358dd'
grant_type = 'client_credentials'
scope = 'InvoicingAPI'

# Prepare the data for the POST request
data = {
    'client_id': client_id,
    'client_secret': client_secret,
    'grant_type': grant_type,
    'scope': scope
}

request_url = f"{token_url}?grant_type={grant_type}&client_id={client_id}&client_secret={client_secret}&scope={scope}"
print("Request URL:", request_url)

# Make the POST request to get the token
response = requests.post(token_url, data=data)

# Check if the request was successful
if response.status_code == 200:
    token = response.json().get("access_token")
    print("Access Token:", token)
else:
    print("Failed to get token:", response.status_code, response.text)
```
